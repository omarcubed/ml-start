```python
class StudentGrades:
  def __init__(self, students=None):
    self.students = students if students is not None else []

  def add_student(self, name, score):
    self.students.append({"name": name, "score": score})

  def get_top_students(self, min_score):
    passed = filter(lambda x: x["score"] > min_score, self.students)
    return sorted(passed, key=lambda x: x["score"], reverse=True)

data = [
    {"name": "Алихан", "score": 75},
    {"name": "Айгерим", "score": 92},
    {"name": "Данияр", "score": 60},
    {"name": "Мадина", "score": 88}
]
journal = StudentGrades(data)
journal.add_student("Тимур", 95)
top = journal.get_top_students(80)
for student in top:
    print(f"{student['name']}: {student['score']}")
```
